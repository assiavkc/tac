# APIs: requêtes HTTP

## Imports

```python
import json
import requests
```

## Utiliser [Wikidata](https://www.wikidata.org/w/api.php) pour chercher des informations sur des journalistes


```python
journaliste = "Hélène Burniaux"

"""   """
headers = {
    "User-Agent": "tactp1 (contact: assia.vkc@gmail.com)"
}
url = "https://www.wikidata.org/w/api.php"
data = {
    "action": "wbsearchentities",
    "search": journaliste,
    "language": "fr",
    "format": "json"}

resp = requests.get(url, data, headers=headers)
try:
    json_list = json.loads(resp.text)
except json.JSONDecodeError: # response is not in JSON as expected, probably the request is being blocked
    print(resp.text)
for item in json_list["search"]:
    nom = item["label"]
    identifiant = item["id"]
    description = item["description"]

    print(f"{nom}  - {identifiant} - {description} ")
```
